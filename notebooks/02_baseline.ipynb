```python
df = pd.read_csv("../data/cleaned_malicious_phish.csv")

label_map = {"benign": 0, "phishing": 1, "malware": 1, "defacement": 1}
df["label"] = df["type"].map(label_map)
```

```python
import tldextract
```

```python
def get_domain(url):
    ext = tldextract.extract(url)
    return f"{ext.domain}.{ext.suffix}"
```

```python
print(get_domain("en.wikipedia.org/wiki/Cat"))    # -> wikipedia.org
print(get_domain("de.wikipedia.org/wiki/Hund"))   # -> wikipedia.org
print(get_domain("secure.paypal.com/login"))      # -> paypal.com
print(get_domain("login.paypal.com/reset"))       # -> paypal.com
```

```python
df["domain"] = df["url"].apply(get_domain)
```

```python
from sklearn.model_selection import GroupShuffleSplit

X = df_features[FEATURE_COLUMNS]
y = df["label"]
groups = df["domain"]

gss = GroupShuffleSplit(n_splits=1, test_size=0.2, random_state=42)

for train_idx, test_idx in gss.split(X,y,groups):
    X_train = X.iloc[train_idx]
    X_test = X.iloc[test_idx]
    y_train = y.iloc[train_idx]
    y_test = y.iloc[test_idx]
    
print(X_train.shape, X_test.shape)


```
